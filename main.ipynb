```python
def linear_contrast_stretching(img):
    return ((img - img.min()) / (img.max() - img.min()) * 255).astype(np.uint8)
```

```python
oif_results = []
bands = {
    1: B1img,
    2: B2img,
    3: B3img,
    4: B4img,
    5: B5img,
    6: B6img,
    7: B7img,
    9: B9img,
    10: B10img,
    11: B11img
}

all_combos = list(combinations(range(1, 10), 3))
filtered_combos = [combo for combo in all_combos if 8 not in combo]

for combo in filtered_combos:

    b1, b2, b3 = combo

    A = bands[b1].astype(np.float32)
    B = bands[b2].astype(np.float32)
    C = bands[b3].astype(np.float32)

    std1 = np.std(A)
    std2 = np.std(B)
    std3 = np.std(C)

    r12 = np.corrcoef(A.flatten(), B.flatten())[0, 1]
    r13 = np.corrcoef(A.flatten(), C.flatten())[0, 1]
    r23 = np.corrcoef(B.flatten(), C.flatten())[0, 1]

    numerator = std1 + std2 + std3
    denominator = abs(r12) + abs(r13) + abs(r23)

    if denominator == 0:
        oif = np.inf
    else:
        oif = numerator / denominator

    oif_results.append((combo, oif))

oif_results_sorted = sorted(oif_results, key=lambda x: x[1], reverse=True)
best_combo, max_oif = oif_results_sorted[0]
band_index1, band_index2, band_index3 = best_combo

try:
    img1 = bands[band_index1]
    img2 = bands[band_index2]
    img3 = bands[band_index3]
except KeyError as e:
    print(f"Error: band {e} not found. Please ensure all bands are loaded correctly.")
    exit()

output_height, output_width = img1.shape
result_OIF_img              = np.zeros((output_height, output_width, 3), dtype=np.float32) 

result_OIF_img = np.dstack((linear_contrast_stretching(img1), linear_contrast_stretching(img2), linear_contrast_stretching(img3)))

plt.imshow(result_OIF_img)
plt.title(f'Best OIF Combination: Bands {best_combo}\nOIF = {max_oif:.2f}')
```

```python
r = 0
RMSE = inf
df_clean = df.copy()
while RMSE > 0.5:

    ones_matrix = np.ones(len(df_clean.u))
    A            = np.column_stack((ones_matrix, df_clean.x, df_clean.y, 
                                    (df_clean.x * df_clean.y), df_clean.x**2, 
                                    df_clean.y**2))

    x_u, _, _, _ = np.linalg.lstsq(A, df_clean.u, rcond=None)
    x_v, _, _, _ = np.linalg.lstsq(A, df_clean.v, rcond=None)

    l_pr_u = A @ x_u
    l_pr_v = A @ x_v

    du = l_pr_u - df_clean.u
    dv = l_pr_v - df_clean.v

    d = np.sqrt(du**2 + dv**2)
    RMSE = np.sqrt(np.mean(((du)**2) + ((dv)**2)))

    df_clean["du"] = du
    df_clean["dv"] = dv
    df_clean["d"] = d


    df_clean.drop(df_clean.sort_values("d", ascending=False).index[0], inplace = True)

    r += 1
    print(f"*********ROUND {r}*************\n")
    print(f"Current RMSE = {RMSE}")
    
print(f"Number Of Remaining Points: {df_clean.shape[0]}\nFinal RMSE:\n{round(RMSE, 5)}\nFinal Data:")
df_clean
```
